```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy.optimize import curve_fit


# Define the file path
# Change location to your drive
file_path = '/content/drive/MyDrive/Colab Notebooks/sift2.csv'

# Read the CSV file
df_sit2 = pd.read_csv(file_path)

y_column = df_sit2.columns[0]
x_column = df_sit2.columns[1]

# Apply the scaling to the data in place for the first column
df_sit2[y_column] = df_sit2[y_column].astype(float) / 1000

print(f"Successfully loaded '{file_path}'. First column ('{y_column}') scaled by 1/1000. Here are the first 5 rows:")
display(df_sit2.head())
```

```python
y_column = df_sit2.columns[0] #column 1: data
x_column = df_sit2.columns[1] #column 2: event number.


#This graph gives us no useful information
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_sit2, x=x_column, y=df_sit2[y_column])
plt.title('Scatter Plot of First Column vs. Second Column')
plt.xlabel(f'Column: {x_column}')
plt.ylabel(f'Column: {y_column} (already scaled)') # Update ylabel
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
